```python
import random
from pyspark.sql import functions as  F
```

```python
def generate_key():
    r=random.random()
    if r<0.90 :
        return 'A'
    else:
        return random.choice(['A','B','C','D','E','F','G'])
    
```

```python
gen_key=F.udf(generate_key)
```

```python
df=spark.range(0,100000).withColumn("key",gen_key()).withColumn("values",F.random()*1000)
```

```python
df.groupBy("key").count().show()
```
